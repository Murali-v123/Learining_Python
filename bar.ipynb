```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[6,7,8,9,11]

plt.bar(x,y,color="yellow")
plt.title("bar plot")
plt.xlabel("scorea")
plt.ylabel("scoreb")
for i in range(len(x)):
    plt.text(x[i], y[i]+1, str(y[i]), ha='center', va='bottom')
plt.ylim(0,max(y)+2)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x=[1,2,3,4,5]
y=[6,7,8,9,11]
z=[7,8,9,10,12]

width=0.4
a=np.arange(len(x))
plt.bar(a-width/2,y,width,label="y")
plt.bar(a+width/2,z,width,label="z")
plt.title("bar plot")
plt.xlabel("scorea")
plt.ylabel("scoreb")
```

```python
import matplotlib.pyplot as plt
import numpy as np
x=[1,2,3,4,5]
y=[6,7,8,9,11]
z=[7,8,9,10,12]

width=0.4
a=np.arange(len(x))
plt.barh(a-width/2,y,width,label="y")
plt.barh(a+width/2,z,width,label="z")
plt.title("bar plot")
plt.xlabel("scorea")
plt.ylabel("scoreb")
plt.savefig("bar_plot.png")
```
