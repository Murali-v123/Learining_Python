```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
score=np.random.normal(70,10,100)
```

```python
plt.hist(score,color="purple",edgecolor="black",bins=10)
plt.ylabel("Scores")
plt.xlabel("#students")
```

```python
import matplotlib.pyplot as plt
number1 = [1, 2.2, 2, 3, 3, 3, 4]        
number2 = [6, 2.6, 7, 8, 8, 3.8, 5, 4.10]

plt.hist(number1,color="red", alpha=0.6, edgecolor="black", label="Group 1")
plt.hist(number2,color="green", alpha=0.6, edgecolor="black", label="Group 2")

plt.legend()
plt.grid()
plt.show()
```
