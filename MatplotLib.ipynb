```python
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[5,6,7,8]
z=[5,5,3,9]
plt.xlabel("num")
plt.ylabel("num")
plt.title("title")
plt.plot(x,y,"o--y",label="num1")
plt.plot(x,z,"4--g",label="num1")
# plt.legend(loc="upper right")
plt.legend(loc="upper left")
```

```python
x=[1,2,3,4]
y=[5,6,7,8]
z=[5,5,3,9]
plt.xlabel("num")
plt.ylabel("num")
plt.title("title")
plt.plot(x,y,color="yellow",linewidth=3,label="num1")
plt.plot(x,z,marker="4",linestyle="--",color="red",label="num1")
# plt.legend(loc="upper right")
plt.legend(loc="upper left")
```
