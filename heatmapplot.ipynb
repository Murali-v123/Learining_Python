```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
flights=sns.load_dataset("flights")
```

```python
flights.head()
```

```python
flight_pivot=flights.pivot(index="month",columns="year",values="passengers")
```

```python
sns.heatmap(
    flight_pivot,
    cmap="coolwarm",
    annot=True,
    fmt="d"
)

plt.tight_layout()
plt.title("Passengers count")
```
