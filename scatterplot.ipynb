```python
import matplotlib.pyplot as plt
```

```python
a=[1,2,3,4,3,5]
c=[6,5,7,8,4,3]
plt.scatter(a,c)
plt.title("scatter plot")
plt.grid()
```

```python
a=[1,2,3,4,3,5]
c=[6,5,7,8,4,3]
colors=["green","yellow","red","purple","red","red"]
plt.scatter(a,c,color=colors,marker="<")
plt.title("scatter plot")
plt.grid()
```

```python
a=[1,2,3,4,3,5]
c=[6,5,7,8,4,3]
colors=["green","yellow","red","purple","red","red"]
plt.scatter(a,c,color=colors,marker="o",alpha=0.5)
plt.title("scatter plot")
plt.grid()
```

```python
a=[1,2,3,4,3,5]
c=[6,5,7,8,4,3]
e=["a","c","d","e","f","g"]
# plt.scatter(a,c,cmap="OrRd",c=c)
plt.scatter(a,c,cmap="plasma",c=c)
plt.title("scatter plot")
plt.grid()
plt.colorbar(label="c")
for i in range(len(e)):
    plt.annotate(e[i],xy=(a[i],c[i]),xytext=(a[i],c[i]))
```
