```python
url = "https://toscrape.com"
response = requests.get(url)
html_content = response.text
```

```python
page_count=1
while True:
    URL=f"https://toscrape.com{page_count}/"
    res=requests.get(URL)

    soup=BeautifulSoup(res.text,"lxml")
    quotes=soup.select("div.quote")

    if not quotes:
        print("invalid")
        break

    with open(f"scraped_data/{page_count}.html","w") as f:
        f.write(res.text)
        print(page_count)

    page_count=page_count+1
```
