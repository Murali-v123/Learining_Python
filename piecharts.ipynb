```python
import matplotlib.pyplot as plt
a=[1,2,3,4,5]
b=["one","two","three","four","five"]

plt.pie(a,labels=b)
```

```python
a=[1,2,3,4,5]
b=["one","two","three","four","five"]
plt.style.use("dark_background")
plt.pie(a,labels=b,autopct="%1.1f%%")
# plt.style.available
```

```python
a=[1,2,3,4,5]
b=["one","two","three","four","five"]
plt.style.use("dark_background")
color=["red","blue","orange","pink","purple"]
plt.pie(a,labels=b,autopct="%1.1f%%",colors=color,
       wedgeprops={
           "edgecolor":"white",
           "linewidth":2,
           "linestyle":"--",
       }
       )
```

```python
a=[1,2,3,4,5]
b=["one","two","three","four","five"]
# plt.style.use("dark_background")
color=["red","blue","orange","pink","purple"]
plt.pie(a,labels=b,autopct="%1.1f%%",colors=color,
       wedgeprops={
           "edgecolor":"white",
           "linewidth":2,
           "linestyle":"--",
           "fill":False
       }
       )
```

```python
a=[1,2,3,4,5]
b=["one","two","three","four","five"]
# plt.style.use("dark_background")
explode=[0,0,0.2,0,0.1]
color=["red","blue","orange","pink","purple"]
plt.pie(a,labels=b,autopct="%1.1f%%",colors=color,
       wedgeprops={
           "edgecolor":"white",
           "linewidth":1,
           "linestyle":"--",
       },
        explode=explode
       )
```

```python
a=[1,2,3,4,5]
b=["one","two","three","four","five"]
plt.style.use("default")
# explode=[0,0,0.2,0,0.1]
color=["red","blue","orange","pink","purple"]
plt.pie(a,labels=b,autopct="%1.1f%%",colors=color,
       wedgeprops={
           "edgecolor":"white",
           "linewidth":1,
           "linestyle":"--",
       },
        # explode=explode,
        startangle=130,
        shadow=True
       )
```
